```python
# 2.1
import torch

# 创建两个 5×5 的随机张量 A、B
A = torch.rand(5, 5)
B = torch.rand(5, 5)

# 创建一个与 A、B 形状和 dtype 相同的零张量 C
C = torch.zeros_like(A)

# 各种运算
add_result = A + B                    # 元素加法
sub_result = A - B                    # 元素减法
mul_result = A * B                    # 元素逐项相乘
matmul_result = A @ B                 # 矩阵乘法（等价于 torch.matmul(A, B)）

# 打印结果
print("Tensor A:\n", A)
print("Tensor B:\n", B)
print("Zero Tensor C:\n", C)
print("A + B:\n", add_result)
print("A - B:\n", sub_result)
print("A * B (element-wise):\n", mul_result)
print("A @ B (matrix multiplication):\n", matmul_result)
```

```python
# 2.3
import torch

# 创建一个 3×3 的张量，并启用自动求导
X = torch.rand(3, 3, requires_grad=True)
print("Original Tensor X:\n", X)

# 定义另一个常量矩阵
M = torch.tensor([[1., 0., 2.],
                  [1., 1., 0.],
                  [0., 3., 1.]])

# 元素乘法（逐元素相乘）
Y = X * M

# 求和
S = Y.sum()

# 反向传播
S.backward()

# 输出梯度
print("\nResult after element-wise multiplication:\n", Y)
print("\nSum of elements (scalar output):", S.item())
print("\nGradient of S with respect to X:\n", X.grad)
```

```python
# 2.4
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
import matplotlib.pyplot as plt
import numpy as np

# 1. 数据准备（使用MNIST数据集）
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.1307,), (0.3081,))  # MNIST标准化
])

# 加载训练集（仅用训练集即可，无需测试集）
train_dataset = datasets.MNIST(
    root='./data', train=True, download=True, transform=transform
)
train_loader = torch.utils.data.DataLoader(
    train_dataset, batch_size=64, shuffle=True
)

# 2. 定义多层全连接网络
class SimpleMLP(nn.Module):
    def __init__(self, layers=10, hidden_dim=128):
        super().__init__()
        # 输入层（28*28=784）→ 多层隐藏层 → 输出层（10类）
        self.layers = nn.ModuleList()
        input_dim = 784
        for _ in range(layers):
            self.layers.append(nn.Linear(input_dim, hidden_dim))
            self.layers.append(nn.ReLU())  # 用ReLU激活
            input_dim = hidden_dim
        self.fc_out = nn.Linear(hidden_dim, 10)  # 输出层

    def forward(self, x):
        x = x.view(-1, 784)  # 展平图像
        for layer in self.layers:
            x = layer(x)
        x = self.fc_out(x)
        return x

# 3. 初始化组件
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = SimpleMLP(layers=10).to(device)  # 4层隐藏层（共5层可训练层）
criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)  # 无动量SGD，避免干扰

# 4. 训练并记录梯度
gradient_stats = []  # 存储各层梯度绝对值均值（格式：[epoch][layer] = 均值）

for epoch in range(10):  # 训练10个epoch
    model.train()
    epoch_gradients = []  # 记录当前epoch各层的梯度均值
    
    for batch_idx, (data, target) in enumerate(train_loader):
        data, target = data.to(device), target.to(device)
        optimizer.zero_grad()  # 清零梯度
        output = model(data)
        loss = criterion(output, target)
        loss.backward()  # 反向传播计算梯度
        
        # 只记录第一个batch的梯度（避免冗余，趋势一致）
        if batch_idx == 0:
            # 遍历所有可训练层（Linear层）
            layer_idx = 0
            for module in model.modules():
                if isinstance(module, nn.Linear):
                    # 计算当前层权重梯度的绝对值均值
                    grad_mean = module.weight.grad.abs().mean().item()
                    epoch_gradients.append(grad_mean)
                    layer_idx += 1
            break  # 只处理第一个batch
    
    gradient_stats.append(epoch_gradients)
    print(f'Epoch {epoch+1}/10, 梯度均值（从浅到深）: {[f"{g:.6f}" for g in epoch_gradients]}')

# 5. 可视化结果
# 计算所有epoch的梯度均值（按层）
avg_gradients = np.mean(gradient_stats, axis=0)
layers = np.arange(1, len(avg_gradients)+1)  # 层索引（1,2,3,...）

plt.figure(figsize=(8, 5))
plt.plot(layers, avg_gradients, 'o-', color='b', linewidth=2)
plt.xlabel('layers')
plt.ylabel('avg_gradients')
plt.title('comparison')
plt.grid(alpha=0.3)
plt.xticks(layers)
plt.show()
```

2.4
The closer a layer is to the output, the larger its gradient; conversely, the closer it is to the input, the smaller the gradient becomes. This indicates that under the current network depth and initialization settings, there is still a certain degree of gradient attenuation. By lowering the learning rate, introducing Batch Normalization, using ReLU or LeakyReLU activations, and adopting proper parameter initialization, the gradient distribution across layers can be made more balanced, thereby improving the trainability of the earlier layers.

```python
# 2.5
import torch
import torch.nn as nn
import torch.optim as optim

# 1) 定义一个最简单的线性回归模型：y = wx + b
class LinearRegressionModel(nn.Module):
    def __init__(self):
        super().__init__()
        # in_features=1, out_features=1
        self.linear = nn.Linear(1, 1)

    def forward(self, x):
        return self.linear(x)

# 2) 造一点“假数据” y = 2x + 3 + 噪声
torch.manual_seed(0)
N = 100
x = torch.randn(N, 1)
true_w, true_b = 2.0, 3.0
y = true_w * x + true_b + 0.1 * torch.randn(N, 1)

# 3) 实例化模型、损失、优化器
model = LinearRegressionModel()
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.1)

# 4) 训练 1 个 epoch（你也可以多跑几次）
model.train()
for epoch in range(20):   # 改成 50 看得更明显
    # 前向
    y_pred = model(x)
    loss = criterion(y_pred, y)

    # 反向
    optimizer.zero_grad()
    loss.backward()

    # 参数更新
    optimizer.step()

    print(f"epoch: {epoch}, loss: {loss.item():.6f}")

# 5) 保存模型权重
save_path = "linreg.pt"
torch.save(model.state_dict(), save_path)
print("weights saved to", save_path)

# 6) 新建同结构的模型，并加载权重
new_model = LinearRegressionModel()
new_model.load_state_dict(torch.load(save_path))
new_model.eval()

# 7) 用新模型做一次推理
test_x = torch.tensor([[0.0], [1.0], [2.0]])
test_y = true_w * test_x + true_b
with torch.no_grad():
    pred = new_model(test_x)
print("test_x:\n", test_x)
print("test_y:\n", test_y)
print("pred:\n", pred)
```
